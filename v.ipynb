```python
import pandas as pd
df=pd.read_csv(r'C:\Users\DEVA NANTHAN\Documents\amazon_music_project\single_genre_artists.csv')
df.head(10)
```

```python
import numpy as np
```

```python
df = df.drop_duplicates()
```

```python
df['duration_min']=df['duration_ms']/60000
df['duration_log'] = np.log1p(df['duration_ms'])
```

```python
df.describe()
```

```python
featuers= ['danceability','energy','loudness','speechiness','acousticness','instrumentalness','liveness','valence','tempo','duration_log']
x=df[featuers]
```

```python
x.describe()
```

```python
x=x.drop('instrumentalness',axis=1)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
inertia=[]
silhouette_scores=[]
k_range=range(2,12)
for k in k_range:
    kmeans=KMeans(n_clusters=k, random_state=42, max_iter=1000) 
    cluster_labels=kmeans.fit_predict(x_scaled)
    sil_score=silhouette_score(x_scaled, cluster_labels)
    silhouette_scores.append(sil_score)
    inertia.append(kmeans.inertia_)


```

```python
#plt.figure(figsize=(14,16))
plt.subplot(1,2,1)
plt.plot(k_range, inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.xticks(k_range)
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(k_range, silhouette_scores, marker='o', color='orange')
plt.title('Silhouette Scores for Different Values of k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(k_range)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, max_iter=1000)

cluster_labels = kmeans.fit_predict(x_scaled)

cluster_labels
```

```python
df['clusters']=cluster_labels
```

```python
featuers
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=3)
x_pca=pca.fit_transform(x_scaled)
```

```python
cluster_colors = {0: '#1f77b4',  # Blue
                  1: '#ff7f0e',  # Orange
                  2: '#2ca02c',  # Green
                  3: '#d62728',  # Red
                  }  


colors = df['clusters'].map(cluster_colors)

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(projection='3d')

scatter = ax.scatter(x_pca[:,0], x_pca[:,1],x_pca[:,2], 
                     c=colors,  # Use mapped solid colors
                     marker='o')

# ax.set_xlabel('Monetary Value')
# ax.set_ylabel('Frequency')
# ax.set_zlabel('Recency')

ax.set_title('3D Scatter Plot of Customer Data by Cluster')
```

```python
cluster_summary = df.groupby('clusters')[featuers].mean()
print(cluster_summary)
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x=x_pca[:,0], y=x_pca[:,1], hue=df['clusters'], palette=cluster_colors)
plt.title('2D PCA Scatter Plot of Clusters')
plt.show()
```

```python
for i in df['clusters'].unique():
    print(f"Cluster {i}:")
    print(df[df['clusters']==i][['duration_min','energy','loudness']].describe())
    print('**************************************************************************')
```
